```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

text = """X:1
T:Simple Tune
M:4/4
L:1/4
K:C
C D E F | G A B c |
c B A G | F E D C |
"""
chars = sorted(set(text))
to_id = {c: i for i, c in enumerate(chars)}
to_char = {i: c for c, i in to_id.items()}
seq_len = 20
data = [to_id[c] for c in text]
X, y = [], []
for i in range(len(data) - seq_len):
    X.append(data[i:i + seq_len])
    y.append(data[i + seq_len])
X, y = np.array(X), np.array(y)

model = Sequential([Embedding(len(chars), 32), LSTM(64), Dense(len(chars), activation="softmax")])
model.compile(optimizer="adam", loss="sparse_categorical_crossentropy")
model.fit(X, y, epochs=30, verbose=0)
seq, out = data[:seq_len], ""
for _ in range(200):
    pred = model.predict(np.array([seq]), verbose=0)[0]
    nxt = np.random.choice(len(chars), p=pred)
    out += to_char[nxt]
    seq = seq[1:] + [nxt]
with open("generated_music.abc", "w") as f:
    f.write(text[:text.find("C D E F")] + out)
print("ABC file generated")
# Optional system command (abc2midi must be installed):
# !abc2midi generated_music.abc -o generated_music.mid
```
