```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report

path = "UCI HAR Dataset/"
X_train = pd.read_csv(path + "train/X_train.txt", sep=r"\\s+", header=None)
X_test = pd.read_csv(path + "test/X_test.txt", sep=r"\\s+", header=None)
y_train = pd.read_csv(path + "train/y_train.txt", header=None).values.ravel()
y_test = pd.read_csv(path + "test/y_test.txt", header=None).values.ravel()

pd.Series(y_train).value_counts().plot(kind="bar")
plt.show()
sns.heatmap(X_train.iloc[:, :10].corr())
plt.show()

rf = RandomForestClassifier(n_estimators=100, random_state=42).fit(X_train, y_train)
p1 = rf.predict(X_test)
print("RF Accuracy:", accuracy_score(y_test, p1))
svm = SVC().fit(X_train, y_train)
p2 = svm.predict(X_test)
print("SVM Accuracy:", accuracy_score(y_test, p2))
print(classification_report(y_test, p2))
```
